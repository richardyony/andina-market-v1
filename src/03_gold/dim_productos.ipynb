```python
import dlt
from pyspark.sql import functions as F

from comun_gold import leer_silver, sk
```

## 2. Tabla `dim_productos`
Catálogo de productos (estado actual, SCD1).

```python
@dlt.table(name="dim_productos", comment="Catálogo de productos (estado actual, SCD1).")
def dim_productos():
    return leer_silver("productos").select(
        sk("id_producto").alias("sk_producto"), "id_producto", "sku", "nombre_producto", "categoria",
        F.col("precio").alias("precio_actual"), "estado", "tiene_descripcion")
```
